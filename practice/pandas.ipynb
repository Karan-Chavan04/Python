In the AI pipeline, Pandas is your data cleaning powerhouse. Before any data gets converted into tensors for neural networks, it spends most of its time sitting in a Pandas DataFrame being filtered, cleaned, and transformed.

1. Core Architecture: Series & DataFrames

Series (1D): A single labeled column (built on top of a 1D NumPy array).

DataFrame (2D): A table with labeled axes (rows and columns), where each column is a Series.

```python
import pandas as pd 
#creating dataframes from a dictionary
data={
    "age":[22,25,28,None],
    "Score":[88,92,None,75],
    "Passed":[True,True,False,False]
}
df=pd.DataFrame(data)
print(df)
```

2. Essential ML Data Preprocessing Patterns

When preparing data for AI models, these are the top 4 operations you'll run constantly:

A. Handling Missing Data (Imputation)

Real-world datasets are messy. Neural networks will throw errors if they see NaN (Not a Number) values.

```python
#Check Missing Values
print(df.isnull().sum())

#Fill missing numerical values with the mean/median
df["age"]=df["age"].fillna(df["age"].median())
df["Score"]=df["Score"].fillna(df["Score"].median())

print(df)
```

B. Filtering & Querying (Boolean Masking)

Just like NumPy boolean indexing, Pandas lets you filter rows using logical conditions.

```python
# Filter rows where Score > 80
high_scorers = df[df["Score"] > 80]
print(high_scorers)
```

C. Feature Engineering & Mapping

Transforming text categories into numerical data so models can process them.

```python
# Map boolean or categorical strings to 0/1
df["Passed"] = df["Passed"].astype(int)
```

D. Converting DataFrames back to NumPy Arrays

Once your data is cleaned, you convert it back to NumPy arrays (or PyTorch/TensorFlow tensors) for training.

```python
# Extract features as a raw 2D NumPy array
X = df[["age", "Score" , "Passed"]].values
print(X)  # <class 'numpy.ndarray'>
```

Pandas Series (1D)

A series is essentially a 1D Numpy array with an Index (row label)
Unlike NumPy arrays where you access elements only by position (scores[0]), in a Series you can access them by label (scores['Alice']) OR by position (scores.iloc[0]).

```python
import pandas as pd
import numpy as np

# Creating a Series from a list
scores = pd.Series([85, 90, 78, 92], index=['Alice', 'Bob', 'Charlie', 'David'])
print(scores)
```

Pandas Dataframes (2D)

A DataFrame is a 2-dimensional table—think of an Excel sheet or a SQL table. Each column in a DataFrame is actually a Pandas Series.

```python
# Creating a DataFrame from a dictionary of lists
data = {
    "Name": ["Alice", "Bob", "Charlie", "David"],
    "Age": [24, 27, 22, 32],
    "Score": [85, 90, 78, 92],
    "Passed": [True, True, False, True]
}

df = pd.DataFrame(data)
print(df)
```

```python
ai_df = pd.DataFrame({
    "Model_name" : ["Gemini","ChatGPT","Claude","Anthropic"],
    "Parameters_Billions":[70,100,200,300],
    "Accuracy":[90,80,75,95],
    "Is_opensource":[True,True,False,True]
})
print(ai_df)
print(ai_df.info())
acc_series=ai_df["Accuracy"]
print(type(acc_series))
```

Data Selection & Indexing (loc vs iloc)

1. iloc (integer-location based)

Selects data using numerical index positions (0, 1, 2...), exactly like NumPy arrays.

```python
# Format: df.iloc[row_index, column_index]

# Get the first row (Gemini)
print(ai_df.iloc[0])

# Get the first 2 rows, and first 2 columns
print(ai_df.iloc[0:2, 0:2])
```

2. loc (Label-location based)

Selects data using row labels and column names.

```python
# Format: df.loc[row_label, column_name]

# Get all rows, but only 'Model_name' and 'Accuracy' columns
print(ai_df.loc[:, ["Model_name", "Accuracy"]])
```

3.Filtering Data (Boolean Indexing)

This is where Pandas shines for AI dataset preparation. You can filter rows based on conditions.

```python
# Filter models where Accuracy is >= 80
high_acc = ai_df[ai_df["Accuracy"] >= 80]
print(high_acc)
print()
# Combine conditions with & (AND) or | (OR)
# Filter models that are Open Source AND have > 80 Accuracy
top_open_source = ai_df[(ai_df["Is_opensource"] == True) & (ai_df["Accuracy"] > 80)]
print(top_open_source)
```

```python
print(ai_df.loc[[1,2],["Model_name","Parameters_Billions"]])
print()
print(ai_df.loc[ai_df["Parameters_Billions"]>90])
```

Importing and exporting data in Pandas

we will check for few most commonly used file formats

1. Comma Seperated Values (CSV)

CSVs are the most common data format in Machine Learning

```python
import pandas as pd

ai_df.to_csv('dataset.csv',index=False )
# 📥 IMPORT (Reading a CSV)
df = pd.read_csv('dataset.csv')
print(df)
df = pd.read_csv('dataset.csv',usecols=["Model_name","Accuracy"]) #Loads only the specified columns
print(df)
# 📤 EXPORT (Saving to a CSV)
# Always use index=False, otherwise Pandas will save row numbers (0, 1, 2...) as a column!
df.to_csv('cleaned_dataset.csv', index=False)
print(df)
```

2. Excel Files (.xlsx)

Pandas works seamlessly with Excel spreadsheets (requires the openpyxl library under the hood).

3. JSON Files (.json)

JSON is standard when fetching data from Web APIs or working with unstructured text data in Generative AI pipelines.

4. Converting directly to/from NumPy & Python Dicts

When moving data into deep learning frameworks (PyTorch or TensorFlow), you'll often convert DataFrames into NumPy arrays or Python dictionaries.

```python
# DataFrame -> NumPy Array (for model input)
X_array = ai_df[['Parameters_Billions', 'Accuracy']].values
print(type(X_array)) 
# DataFrame -> Python Dictionary
data_dict = ai_df.to_dict(orient='records')
print(type(data_dict))
# Python Dictionary -> DataFrame
df_new = pd.DataFrame(data_dict)
print(type(df_new))
```
